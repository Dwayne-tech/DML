```python
import matplotlib.pyplot as plt
import numpy as np

# Generate sample data
x = np.random.randn(1000)
y = np.random.randn(1000)

# Create a scatterplot
plt.scatter(x, y, color='yellow', alpha=0.7)

# Customize the plot
plt.title('Scatterplot of Random Data')
plt.xlabel('X Values')
plt.ylabel('Y Values')
plt.grid(True)

# Show the plot
plt.show()
```
