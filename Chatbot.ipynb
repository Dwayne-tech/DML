```python

# Get the most relevant sentence
def get_most_relevant_sentence(query, sentences, vectorizer):
    # Preprocess the query
    query_processed = preprocess(query)
    query_processed = ' '.join(query_processed)
    # Vectorize the query
    query_vec = vectorizer.transform([query_processed])
    # Compute similarities
    similarities = cosine_similarity(query_vec, vectorizer.transform(sentences))
    # Get the index of the most similar sentence
    most_similar_idx = similarities.argmax()
    return sentences[most_similar_idx]
```

```python
# Chatbot function
def chatbot(user_input, data, vectorizer):
    response = get_most_relevant_sentence(user_input, data, vectorizer)
    return response
```

```python
# Main function to create the Streamlit app
def main():
    st.title('Investing Chatbot')
    st.write('Ask questions about investing and money management.')

    # Load data and vectorizer
    data = load_data('/content/How to invest book.txt')
    vectorizer = TfidfVectorizer()
    vectorizer.fit_transform(data)

    # Get user input
    user_input = st.text_input('You:', '')
    if user_input:
        # Get chatbot response
        response = chatbot(user_input, data, vectorizer)
        st.text_area('Chatbot:', value=response, height=100, max_chars=None)

if __name__ == '__main__':
    main() # Indented this line to be part of the 'if' block
```
