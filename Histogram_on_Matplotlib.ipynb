```python
import matplotlib.pyplot as plt
import numpy as np

# Generate sample data
data = np.random.randn(1000)

# Create a histogram
plt.hist(data, bins=30, density=True, color='blue', alpha=0.7)

# Customize the plot
plt.title('Histogram of Random Data')
plt.xlabel('Value')
plt.ylabel('Probability Density')
plt.grid(True)

# Show the plot
plt.show()
```
