```python
import random

questions = [
    {
        "prompt": "Which of the following is the largest planet in our solar system?",
        "options": ["Earth", "Mars", "Jupiter", "Saturn"],
        "correct_answer": "Jupiter"
    },
    {
        "prompt": "What is the capital of France?",
        "options": ["Paris", "London", "Berlin", "Madrid"],
        "correct_answer": "Paris"
    },
    {
        "prompt": "Which mountain is the highest in Africa?",
        "options": ["Mount Kilimanjaro", "Mount Kenya", "Mount Ruwenzori", "Mount Elgon"],
        "correct_answer": "Mount Kilimanjaro"
    },
    {
        "prompt": "Which of the following is the largest country in the world by land area?",
        "options": ["Russia", "China", "Canada", "United States", "Brazil"],
        "correct_answer": "Russia"
    }
]

def shuffle_questions(questions):
    """Shuffles the order of questions randomly."""
    random.shuffle(questions)

def display_question(question):
    """Displays a question and its options to the user."""
    print(question["prompt"])
    for i, option in enumerate(question["options"]):
        print(f"{i+1}. {option}")

def get_user_input(question):
    """Prompts the user for an answer and validates it."""
    while True:
        try:
            answer = int(input("Enter your answer (1-4): "))
            if 1 <= answer <= len(question["options"]):
                return answer
            else:
                print("Invalid input. Please enter a number between 1 and", len(question["options"]))
        except ValueError:
            print("Invalid input. Please enter a number.")

def check_answer(question, answer):
    """Checks if the user's answer is correct and provides feedback."""
    if question["options"][answer - 1] == question["correct_answer"]:
        print("Correct!")
        return 1
    else:
        print("Incorrect. The correct answer is:", question["correct_answer"])
        return 0

def play_quiz():
    """Plays the quiz and calculates the final score."""
    score = 0
    shuffle_questions(questions)

    for question in questions:
        display_question(question)
        answer = get_user_input(question)
        score += check_answer(question, answer)

    print("Quiz complete!")
    print("Your final score is:", score, "out of", len(questions))
    print("Thank you for playing!")

if __name__ == "__main__":
    play_quiz()
```
